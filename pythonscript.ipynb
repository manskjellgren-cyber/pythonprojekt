```python
# ========================================
# CELL 1: UTAN OOP vs MED OOP
# ========================================
# Jämför hur man hanterar datapunkter för AI-träning
# ========================================

print("=" * 60)
print("📊 UTAN OOP vs MED OOP")
print("=" * 60)
print("Exempel: Hantera datapunkter för en AI-modell (temperatur → regn eller sol)")
print("-" * 60)

# ========================================
# DEL 1: UTAN OOP (dåligt)
# ========================================
print("\n❌ UTAN OOP:")

# 3 datapunkter kräver 9 variabler!
temp1 = 22.5
humidity1 = 65
label1 = 0          # 0 = sol

temp2 = 23.0
humidity2 = 70
label2 = 1          # 1 = regn

temp3 = 21.5
humidity3 = 80
label3 = 1          # 1 = regn

print("   Data:")
print(f"      Punkt 1: temp={temp1}°, fukt={humidity1}%, label={label1} (sol)")
print(f"      Punkt 2: temp={temp2}°, fukt={humidity2}%, label={label2} (regn)")
print(f"      Punkt 3: temp={temp3}°, fukt={humidity3}%, label={label3} (regn)")

# Försök beräkna medeltemperatur
medel_temp = (temp1 + temp2 + temp3) / 3
print(f"\n   📈 Medeltemperatur: {medel_temp}°C")

print("\n   🔴 PROBLEM:")
print("      - 1000 punkter → 3000 variabler!")
print("      - Lätt att blanda ihop variabler")
print("      - Svårt att lägga till ny feature (t.ex. vind)")
print("      - Kod blir oöverskådlig")

# ========================================
# DEL 2: MED OOP (bra)
# ========================================
print("\n" + "=" * 60)
print("✅ MED OOP:")
print("=" * 60)

# Definiera klassen EN gång
class DataPoint:
    def __init__(self, temp, humidity, label):
        self.temp = temp
        self.humidity = humidity
        self.label = label

    def predict(self):
        """Simulerar AI-modellens prediktion"""
        if self.label == 1:
            return "regn"
        else:
            return "sol"

# Skapa objekt – EN rad per datapunkt!
points = [
    DataPoint(22.5, 65, 0),
    DataPoint(23.0, 70, 1),
    DataPoint(21.5, 80, 1)
]

print("\n   📋 Alla datapunkter i EN lista:")
for i, point in enumerate(points):
    print(f"      Punkt {i+1}: temp={point.temp}°, fukt={point.humidity}% → {point.predict()}")

# Beräkna medeltemperatur – enkel loop!
medel_temp = sum(p.temp for p in points) / len(points)
print(f"\n   📈 Medeltemperatur: {medel_temp}°C")

print("\n   ✅ FÖRDELAR:")
print("      - 1000 punkter → 1000 objekt (inte 3000 variabler)")
print("      - Data samlat i objekt: point.temp, point.humidity")
print("      - Lätt att lägga till ny feature (ändra på 1 ställe)")
print("      - Kod blir ren och lättläst")

# ========================================
# JÄMFÖRELSE – TABELL
# ========================================
print("\n" + "=" * 60)
print("📊 JÄMFÖRELSE: 1000 DATAPUNKTER")
print("=" * 60)

print("""
┌────────────────────┬──────────────────────┬──────────────────────┐
│                    │ UTAN OOP (❌)         │ MED OOP (✅)         │
├────────────────────┼──────────────────────┼──────────────────────┤
│ Antal variabler    │ 3000                 │ 1000 objekt          │
├────────────────────┼──────────────────────┼──────────────────────┤
│ Lägga till feature │ Ändra på 1000 ställen│ Ändra på 1 ställe    │
│ (t.ex. vind)       │                      │ (i klassen)          │
├────────────────────┼──────────────────────┼──────────────────────┤
│ Beräkna medel      │ Skriv formel med     │ Enkel loop:          │
│                    │ 1000 variabler       │ sum(p.temp)/len(p)   │
├────────────────────┼──────────────────────┼──────────────────────┤
│ Kodlängd           │ 3000+ rader          │ ~20 rader            │
└────────────────────┴──────────────────────┴──────────────────────┘
""")

# ========================================
# EXTRA: Svenska AI-bolag använder OOP
# ========================================
print("\n" + "=" * 60)
print("🏢 SÅ HÄR ANVÄNDER SVENSKA AI-BOLAG OOP")
print("=" * 60)

print("""
   🚛 EINRIDE:   SensorReading(temp, speed, battery) → is_critical()
   🏥 KRY:       Patient(age, symptoms) → is_urgent()
   🚗 VOLVO:     TestData(speed, brake_temp) → is_safe()
   💻 LOVABLE:   Prompt(user_input) → generate_code()
""")

print("\n🎯 KLAR! OOP gör AI-utveckling enklare och mer strukturerad!")
 
# ========================================
```

```python
# ========================================
# CELL 1: PROBLEMET UTAN ARV
# ========================================
# Visa varför arv behövs – utan arv måste vi upprepa samma kod om och om igen
# ========================================

print("=" * 60)
print("❌ PROBLEMET UTAN ARV")
print("=" * 60)
print("Tänk dig att du ska skapa 3 olika dataset-klasser...")
print("-" * 60)

# ----------------------------------------
# KLASS 1: Dataset för temperaturer
# ----------------------------------------
print("\n📊 KLASS 1: TemperatureDataset")

class TemperatureDataset:
    def __init__(self, data=None):
        # data=None är global variable 
        if data is None:
            self.data = []
        else:
            self.data = data

    def load_data(self, data):
        self.data = data
        print(f"   Laddade {len(data)} temperaturer")

    def get_stats(self):
        if not self.data:
            return {"count": 0, "min": None, "max": None, "average": None}
        return {
            "count": len(self.data),
            "min": min(self.data),
            "max": max(self.data),
            "average": sum(self.data) / len(self.data)
        }

    def __str__(self):
        return f"TemperatureDataset({self.data})"

# Skapa och använd
temps = TemperatureDataset([22.5, 23.0, 21.5])
print(f"   Skapat: {temps}")
print(f"   Statistik: {temps.get_stats()}")

# ----------------------------------------
# KLASS 2: Dataset för bilder (Samma kod igen!)
# ----------------------------------------
print("\n" + "=" * 60)
print("🖼️ KLASS 2: ImageDataset (samma kod upprepas!)")
print("=" * 60)

class ImageDataset:
    def __init__(self, data=None):
        if data is None:
            self.data = []
        else:
            self.data = data

    def load_data(self, data):
        self.data = data
        print(f"   Laddade {len(data)} bilder")

    def get_stats(self):
        if not self.data:
            return {"count": 0, "min": None, "max": None, "average": None}
        return {
            "count": len(self.data),
            "min": min(self.data) if self.data else None,
            "max": max(self.data) if self.data else None,
            "average": None  # Kan inte beräkna medel på bilder!
        }

    def load_images(self, paths):
        """Egen metod för ImageDataset"""
        self.load_data(paths)
        print(f"   📸 Laddade bilder: {paths}")

    def __str__(self):
        return f"ImageDataset({self.data})"

# Skapa och använd
images = ImageDataset() #separat klass
images.load_images(["bild1.jpg", "bild2.jpg"]) #metoder
print(f"   Skapat: {images}")
print(f"   Statistik: {images.get_stats()}")

# ----------------------------------------
# KLASS 3: Dataset för texter (Samma kod IGEN!)
# ----------------------------------------
print("\n" + "=" * 60)
print("📝 KLASS 3: TextDataset (samma kod upprepas IGEN!)")
print("=" * 60)

class TextDataset:
    def __init__(self, data=None):
        if data is None:
            self.data = []
        else:
            self.data = data

    def load_data(self, data):
        self.data = data
        print(f"   Laddade {len(data)} texter")

    def get_stats(self):
        if not self.data:
            return {"count": 0, "min": None, "max": None, "average": None}
        # Räkna totalt antal tecken
        total_chars = sum(len(text) for text in self.data)
        return {
            "count": len(self.data),
            "total_chars": total_chars,
            "avg_chars": total_chars / len(self.data)
        }

    def tokenize(self): #behöver inte ärva
        """Egen metod för TextDataset – dela upp i ord"""
        result = []
        for text in self.data:
            result.append(text.split())
        return result

    def __str__(self):
        return f"TextDataset({self.data})"

# Skapa och använd
texts = TextDataset(["Hej världen", "AI är kul", "Python är bäst"])
print(f"   Skapat: {texts}")
print(f"   Statistik: {texts.get_stats()}")
print(f"   Tokeniserat: {texts.tokenize()}")

# ----------------------------------------
# PROBLEMET – VISUALISERING
# ----------------------------------------
print("\n" + "=" * 60)
print("📊 PROBLEMET – VISUALISERING")
print("=" * 60)

print("""
┌─────────────────────────────────────────────────────────────────────────────┐
│                                                                             │
│   ❌ UTAN ARV (3 separata klasser)                                          │
│                                                                             │
│   ┌─────────────────────┐    ┌─────────────────────┐    ┌─────────────────┐│
│   │ TemperatureDataset  │    │    ImageDataset     │    │   TextDataset   ││
│   ├─────────────────────┤    ├─────────────────────┤    ├─────────────────┤│
│   │ __init__()          │    │ __init__()          │    │ __init__()      ││
│   │ load_data()         │    │ load_data()         │    │ load_data()     ││
│   │ get_stats()         │    │ get_stats()         │    │ get_stats()     ││
│   │ __str__()           │    │ __str__()           │    │ __str__()       ││
│   │                     │    │ load_images()       │    │ tokenize()      ││
│   └─────────────────────┘    └─────────────────────┘    └─────────────────┘│
│                                                                             │
│   🔴 SAMMA KOD upprepas 3 gånger!                                           │
│   🔴 3 klasser = ~60 rader kod                                              │
│   🔴 Lägg till ny metod? → Ändra på 3 ställen                               │
│                                                                             │
└─────────────────────────────────────────────────────────────────────────────┘
""")

# ----------------------------------------
# SAMMANFATTNING
# ----------------------------------------
print("\n" + "=" * 60)
print("🎯 SAMMANFATTNING – PROBLEMET")
print("=" * 60)

print("""
   🔴 3 klasser = mycket upprepad kod (__init__, load_data, get_stats, __str__)
   🔴 Lägga till ny funktion → ändra på 3 ställen
   🔴 Lätt att göra fel när kod dupliceras
   🔴 Svårt att underhålla

   ✅ LÖSNING: ARV! (Se nästa cell)
   ✅ Samla gemensam kod i EN basklass
   ✅ Låt barnklasserna ärva och lägga till EGET beteende
""")

```

```python
# ========================================
# CELL 2: EINRIDE – ARV I PRAKTIKEN
# ========================================
# Einride använder arv för att hantera olika typer av sensorer i sina lastbilar
# ========================================

print("=" * 60)
print("🚛 EINRIDE – ARV I PRAKTIKEN")
print("=" * 60)
print("Einrides lastbilar har flera sensorer: temperatur, hastighet, batteri.")
print("Alla sensorer har samma grundfunktioner – men olika kritiska gränser!")
print("-" * 60)

# ----------------------------------------
# FÖRÄLDERKLASS: Sensor (gemensam för alla sensorer)
# ----------------------------------------
class Sensor:
    """En grundläggande sensor – alla sensorer ärver från denna"""

    def __init__(self, value, timestamp):
        self.value = value      # Sensorvärdet (t.ex. 22.5°C, 80 km/h, 15%)
        self.timestamp = timestamp  # När mätningen gjordes

    def get_value(self):
        """Returnerar sensorns värde"""
        return self.value

    def is_critical(self):
        """Kollar om värdet är kritiskt (basklass – alltid False)"""
        return False

    def __str__(self):
        return f"{self.__class__.__name__}: {self.value}"

# ----------------------------------------
# BARNKLASS 1: TempSensor (ärver från Sensor)
# ----------------------------------------
class TempSensor(Sensor):
    """Temperatursensor – kritisk om temperaturen är för hög"""

    def is_critical(self):
        """Åsidosätter metoden från föräldern"""
        return self.value > 40  # För hög temperatur (>40°C)

# ----------------------------------------
# BARNKLASS 2: SpeedSensor (ärver från Sensor)
# ----------------------------------------
class SpeedSensor(Sensor):
    """Hastighetssensor – kritisk om hastigheten är för hög"""

    def is_critical(self):
        return self.value > 120  # För hög hastighet (>120 km/h)

# ----------------------------------------
# BARNKLASS 3: BatterySensor (ärver från Sensor)
# ----------------------------------------
class BatterySensor(Sensor):
    """Batterisensor – kritisk om batterinivån är för låg"""

    def is_critical(self):
        return self.value < 10  # För låg batterinivå (<10%)

# ----------------------------------------
# TESTA SENSORERNA
# ----------------------------------------
print("\n📊 TEST 1: Temperatursensorer")
print("-" * 40)

temp1 = TempSensor(22.5, "2026-05-17 10:00")
temp2 = TempSensor(45.0, "2026-05-17 10:01")

print(f"   {temp1} → Kritiskt? {temp1.is_critical()} (22.5°C är OK)")
print(f"   {temp2} → Kritiskt? {temp2.is_critical()} (45°C är FÖR HÖGT!)")

print("\n⚡ TEST 2: Hastighetssensorer")
print("-" * 40)

speed1 = SpeedSensor(80, "2026-05-17 10:00")
speed2 = SpeedSensor(130, "2026-05-17 10:01")

print(f"   {speed1} → Kritiskt? {speed1.is_critical()} (80 km/h är OK)")
print(f"   {speed2} → Kritiskt? {speed2.is_critical()} (130 km/h är FÖR HÖGT!)")

print("\n🔋 TEST 3: Batterisensorer")
print("-" * 40)

battery1 = BatterySensor(45, "2026-05-17 10:00")
battery2 = BatterySensor(8, "2026-05-17 10:01")

print(f"   {battery1} → Kritiskt? {battery1.is_critical()} (45% är OK)")
print(f"   {battery2} → Kritiskt? {battery2.is_critical()} (8% är KRITISKT!)")

# ----------------------------------------
# VISA ARVSHIERARKIN
# ----------------------------------------
print("\n" + "=" * 60)
print("📋 ARVSHIERARKI")
print("=" * 60)

print("""
   ┌─────────────────────────────────────────────────────────────────┐
   │                                                                 │
   │                    class Sensor                                 │
   │                  (förälderklass)                                │
   │                                                                 │
   │   ┌─────────────────────────────────────────────────────────┐   │
   │   │  __init__(value, timestamp)                             │   │
   │   │  get_value() → return value                             │   │
   │   │  is_critical() → return False (basklass)                │   │
   │   └─────────────────────────────────────────────────────────┘   │
   │                              │                                   │
   │           ┌──────────────────┼──────────────────┐               │
   │           │                  │                  │               │
   │           ▼                  ▼                  ▼               │
   │   ┌─────────────┐    ┌─────────────┐    ┌─────────────┐         │
   │   │ TempSensor  │    │ SpeedSensor │    │BatterySensor│         │
   │   │   (barn)    │    │   (barn)    │    │   (barn)    │         │
   │   ├─────────────┤    ├─────────────┤    ├─────────────┤         │
   │   │ is_critical │    │ is_critical │    │ is_critical │         │
   │   │ → value>40  │    │ → value>120 │    │ → value<10  │         │
   │   └─────────────┘    └─────────────┘    └─────────────┘         │
   │                                                                 │
   └─────────────────────────────────────────────────────────────────┘
""")

# ----------------------------------------
# SIMULERA EINRIDES SYSTEM
# ----------------------------------------
print("\n" + "=" * 60)
print("🚛 SIMULERING: Einrides sensor-system")
print("=" * 60)

# Samla alla sensorer i en lista
sensors = [
    TempSensor(22.5, "10:00"),
    SpeedSensor(80, "10:00"),
    BatterySensor(45, "10:00"),
    TempSensor(45.0, "10:01"),
    SpeedSensor(130, "10:01"),
    BatterySensor(8, "10:01")
]

print("\n   🔍 Kontrollerar alla sensorer:")
print("-" * 40)

critical_count = 0
for sensor in sensors:
    status = "🔴 KRITISKT" if sensor.is_critical() else "🟢 OK"
    if sensor.is_critical(): #dvs om sensor critical = true
        critical_count += 1
    print(f"   {sensor} → {status}")

print(f"\n   📊 SAMMANFATTNING: {critical_count} av {len(sensors)} sensorer är kritiska!")

# ----------------------------------------
# SAMMANFATTNING – ARV FÖR EINRIDE
# ----------------------------------------
print("\n" + "=" * 60)
print("🎯 SAMMANFATTNING – ARV FÖR EINRIDE")
print("=" * 60)

print("""
   ✅ ALLA sensorer ärver från SAMMA basklass (Sensor)
   ✅ ALLA sensorer har __init__, get_value(), is_critical()
   ✅ VARJE sensortyp har EGEN is_critical() med olika regler

   💡 Utan arv:
      - 3 separata klasser med samma kod upprepad
      - 30 rader kod → 90 rader kod

   💡 Med arv:
      - 1 basklass + 3 barnklasser
      - 30 rader kod totalt
      - Lätt att lägga till nya sensorer (t.ex. GyroSensor)
""")

print("\n🔜 Nästa cell: Arv i dataset (Dataset → ImageDataset)")
 
# ====
```

Övning med Klass och abstrakt arv Smarta Hem

```python
from abc import ABC, abstractmethod

class SmartEnhet(ABC):
    def __init__(self, namn):
        self.namn = namn
    
    @abstractmethod
    def starta(self):
        pass
    
    # En funktion eller metod i Python (ovan rad 9) 
    # kan inte lämnas helt tom utan kodrader under sig. 
    # När du deklarerar den abstrakta metoden 
    # i föräldraklassen måste du lägga till nyckelordet 
    # pass på raden under, som en tom platshållare.
        
    def visa_status(self):
        print(f"{self.namn} är ansluten till nätverket.")
       
class Lampa(SmartEnhet):
    def __init__(self, namn, ljusstyrka):
        super().__init__(namn)
        self.ljusstyrka = ljusstyrka
    
    def starta(self): # Behöver bara self eftersom datat redan är sparat i objektet
        print(f"Tänder {self.namn} med ljusstyrka {self.ljusstyrka}%.")

class Termostat(SmartEnhet):
    def __init__(self, namn, temperatur):
        super().__init__(namn)
        self.temperatur = temperatur
    
    def starta(self):
        print(f"Sätter {self.namn} på {self.temperatur}°C.")    
        

enheter = [
    Lampa("Kökstak", ljusstyrka=80),
    Termostat("Vardagsrum", temperatur=21.5)
    ]

for enhet in enheter:
    enhet.visa_status()
    enhet.starta()
    print("---")
    
```

```python
class InvalidSensorDataError(ValueError):
    """Kastas när en sensor levererar värden utanför tillåtet intervall."""
    pass

class Kylaggregat:
    def __init__(self, temp: float):
        self.temp = temp
    
    # 1. GETTER körs när någon LÄSER värdet t.ex. print(server.temp)
    @property
    def temp(self) -> float:
        return self._temp
    
    # 2. SETTER körs när någon SKRIVER ett nytt värde t.ex. server.temp = 50.0
    @temp.setter
    def temp(self, value: float) -> None:
        if -50 <= value <= 100.0:
            self._temp = value
        else:
            # print(f"❌ FEL: {value}% är en ogiltig temperatur!")
            raise InvalidSensorDataError(f"{value}°C är utanför tillåtet intervall (-50 till 100).")
            # ersatt raise ValueError med InvalidSensorDataError
            # "Fail Fast principen" koden stoppar direkt innan felaktig data sparas i minnet
    
kyl = Kylaggregat(temp=20.0)
print(f"Starttemp: {kyl.temp}°C")  # Bör skriva ut 20.0°C

try:
    # 1. kod som KAN kasta ett fel  
    kyl.temp = 150.0  # Bör trigga ditt felmeddelande!
except InvalidSensorDataError as err:
    # 2. Kod som körs OM ett ValueError kastades
    # # ersatt ValueError med InvalidSensorDataError
    print(f"⚠️ Fångade ett fel: {err}")
# 3. Koden fortsätter hit utan att krasha
print(f"Aktuell temp är fortfarande: {kyl.temp}°C")    
    
print(f"Efter felaktigt försök: {kyl.temp}°C")  # Bör FORTFARANDE vara 20.0°C

kyl.temp = 15.0  # Giltigt värde
print(f"Ny temp: {kyl.temp}°C")  # Bör skriva ut 15.0°C
    
```
